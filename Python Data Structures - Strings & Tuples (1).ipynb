# Strings (Concatenation, Slicing, and Other Methods) :

```python
# String Concatenation
string1='Hello '
print(string1)
```

```python
string2=input('Enter Your Name :')
print(string1 + string2 )
```

```python
string3='Welcome to Python programming'
print(string3)
```

```python
print(string1+string2+ ',' +string3)
```

# String Slicing and Indexing:

```python
full_string= string1+string2+ ','+' ' +string3
print(full_string)
```

```python
# Print the first character of the string.
print(full_string[0])
```

```python
# Print the last character of the string.
print(full_string[-1])
```

```python
#Print the first 5 characters of the string.
print(full_string[0:5])
```

```python
#Print the last 11 characters of the string.
print(full_string[-11:])
```

```python
#Print the string in reverse.
print(full_string[::-1])
```

```python
#Use slicing and print the word “Python” from the existing string.
print(full_string[-18:-12])

```

# String Methods:

```python
strM = 'Python beginner tutorial'
print(strM)
```

```python
#Convert the sentence to uppercase.
print(strM.upper())
```

```python
#Convert the sentence to lowercase.
print(strM.lower())
```

```python
#Use Capitalize and return the sentence to the original input form.
print(strM.capitalize())
```

```python
#Count the total number of occurrences of character ‘t’ in the string.
print(strM.count('t'))
```

```python
# Replace all occurrences of “Python” with “Data Analytics” in the input string
#strM = “Python beginner tutorial.”

strM.replace('Python','Data Analytics')
```

# Tuples (Creation, Modification and Access) :

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
print(tuple1)
print(tuple2)
```

```python
#a. Concatenate the two tuples and store it in “t_combine”
t_combine=tuple1+tuple2
print(t_combine)
```

```python
# b. Repeat the elements of “t_combine” 3 times
[t_combine]*3
```

```python
#c. Access the 3rd element from “t_combine”
print(t_combine[2])
```

```python
#d. Access the first three elements from “t_combine”
print(t_combine[0:3])
```

```python
#e. Access the last three elements from “t_combine”
print(t_combine[-3::])
```
